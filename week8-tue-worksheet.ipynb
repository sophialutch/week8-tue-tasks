# rank2score

```python
def rank2score(rank):
    point = {1:25, 2: 18, 3: 15, 4: 12, 5: 10, 6:8, 7: 6, 8: 4, 9: 2, 10: 1}
    return point.get(rank, 0)
```

##### ^ had to put 0 after rank bc was getting a keyerror before

##### making score be based of my ranking function

```python
df["score"] = df["weekly_rank"].apply(rank2score)
```

##### Guardian: The Lonely and Great God has one of the most famous and well-known soundtracks known for its music + sound

```python

title = "Guardian: The Lonely and Great God"
showdf = df[df["show_title"] == title]
```

##### grouped the show's rows by country and calculated the sum of points. counted the total charting weeks and collected the weekly rankings into a list for each country.

##### also wanted to see, "ok yes how many total weeks but what were the rankings for all of those weeks"

```python
mesh = showdf.groupby("country_name")
totalscore = mesh["score"].sum()
totalweeks = mesh["weekly_rank"].count()
rankingslist = mesh["weekly_rank"].apply(list)
```

##### adding all labels to df and npt forgetting to sort values @ end

```python
summarydf = pd.DataFrame({"totalscore":totalscore, "totalweeks": totalweeks, "rankingslist": rankingslist,})
summarydf = summarydf.sort_values(by="totalscore",ascending=False)
summarydf
```
